# Reproduction 03 — Figure ingredients
Inspect the deterministic arrays behind the paper figures before rendering publication PDFs.

```python
import numpy as np
from rpgeom.ceilings import laguerre_singvals, shape_ratio
from rpgeom.laws import pairwise_agreement
d = 200
ms = np.array([5, 10, 20, 40, 80, 120])
order_curve = np.array([pairwise_agreement(int(m), d) for m in ms])
shape_curve = np.array([shape_ratio(int(m), d) for m in ms])
order_curve, shape_curve
```

```python
print('Laguerre mode shares:', laguerre_singvals(40, 200, 4)**2)
assert np.all(np.diff(order_curve) > 0)
assert np.all(np.diff(shape_curve) > 0)
```
