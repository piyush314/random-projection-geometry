# 04 — Anisotropy and nonlinear excess
Distance-value recovery and arbitrary-feature recovery coincide in isotropy but may separate under unequal spectral retention.

```python
import numpy as np
from rpgeom.ceilings import alpha_m, alpha_pi
spectrum = np.array([4., 1., 1., 1.])
print('optimal value share:', alpha_m(spectrum, m=1))
print('balanced block share:', alpha_pi(lam=[4.,1.], r=[20,40], s=[5,10]))
```

```python
linear = np.sqrt(4/5)
quadratic = np.sqrt((246 + 2*np.sqrt(201))/311)
print('linear witness', linear, 'quadratic lower bound', quadratic)
assert quadratic > linear
```

**Interpretation.** Unequal block retention lets a nonlinear feature reweight observed and omitted components. Equal fractional retention across blocks removes this mechanism.
