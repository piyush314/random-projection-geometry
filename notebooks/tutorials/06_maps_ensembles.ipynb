# 06 — General maps and ensembles
Rank controls optimal noiseless inference; spectral spread controls an unwhitened norm; small singular values control noisy recovery.

```python
import numpy as np
from rpgeom.ensembles import averaging_corr, chain_corr, kantorovich_bounds, stacking_ceiling
print('chain:', chain_corr([150,100], d=300))
print('average:', averaging_corr(100,400), 'stack:', stacking_ceiling(100,400))
print('Kantorovich:', kantorovich_bounds(np.array([1., .5, .2])))
```

```python
from rpgeom.mmse import direction_recovery, quarter_circle_loss
print(direction_recovery(np.array([2.,1.,.1]), eps=.2))
print('square-Gaussian limiting posterior loss:', quarter_circle_loss(.2))
```

**Interpretation.** A full-rank map may be perfectly invertible without noise, poor as an uncorrected metric, and unstable after noise. These are three different claims.
