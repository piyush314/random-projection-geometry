# 01 — Beta–gamma channel and Laguerre modes
An isotropic projected distance is the scalar observation $T=U+V\mapsto U$ with independent chi-squared blocks.

```python
import numpy as np
from rpgeom.channels import beta_gamma_sample
from rpgeom.ceilings import hgr_isotropic, laguerre_singvals, mutual_info
rng = np.random.default_rng(1)
T, U, V = beta_gamma_sample(64, 768, 20_000, rng)
print('sample correlation', np.corrcoef(T, U)[0, 1])
print('HGR ceiling', hgr_isotropic(64, 768))
```

```python
shares = laguerre_singvals(64, 768, kmax=6) ** 2
print('recoverable variance by Laguerre mode:', shares)
print('mutual information:', mutual_info(64, 768), 'nats')
assert np.all(np.diff(shares) < 0)
```

**Interpretation.** The first mode is centered distance and attains the HGR ceiling. Higher-order fluctuation modes decay more quickly, so nonlinear does not mean unconstrained.
