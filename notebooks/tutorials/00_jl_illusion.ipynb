# 00 — The JL illusion
A relative-distance certificate and a ranking certificate measure different scales. This notebook computes both budgets for the same dimensions.

```python
from rpgeom import budget
report = budget(d=10_000, m=200, q=8)
print(report)
```

```python
assert report.hgr_ceiling < 0.15
assert report.kendall_tau < 0.1
report.to_dict()
```

**Interpretation.** The large distance baseline may meet a useful relative-error target even though the fluctuation-level ranking signal is weak. This is not a failure of JL; it is a mismatch between certificate and decision.
