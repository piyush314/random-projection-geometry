# 02 — Arcsine law and Kendall correlation
The exact pairwise survival probability averages a Gaussian sign law over a beta-distributed projected length.

```python
import numpy as np
from rpgeom.laws import pairwise_agreement, kendall_tau
d = 200
rows = [(m, pairwise_agreement(m, d), kendall_tau(m, d)) for m in (5, 20, 50, 100)]
for row in rows: print(row)
```

```python
for m, p, tau in rows:
    assert abs(tau - (2*p - 1)) < 1e-12
    print(m, 'small-ratio approximation:', 0.5 + np.sqrt(m/d)/np.pi)
```

**Interpretation.** Pairwise agreement approaches chance at order $\sqrt{m/d}$ above one half. Kendall tau removes the chance baseline and makes the collapse visible.
