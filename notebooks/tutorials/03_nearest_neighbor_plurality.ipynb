# 03 — Nearest neighbors and plurality stability
A common argmin is a joint event; pairwise probabilities cannot simply be multiplied because candidates share a query.

```python
import numpy as np
from rpgeom.laws import plurality_kernel, plurality_kernel_mc
rho, q = np.sqrt(20/10_000), 8
exact = plurality_kernel(rho, q)
mc, se = plurality_kernel_mc(rho, q, n=50_000, rng=np.random.default_rng(3))
print('quadrature', exact, 'Monte Carlo', mc, '+/-', se, 'chance', 1/q)
```

```python
assert abs(mc-exact) < 5*se
for candidates in (2, 5, 10):
    print(candidates, plurality_kernel(0.25, candidates), 'chance', 1/candidates)
```

**Interpretation.** The fixed-candidate limit separates the compression signal from the chance baseline. Growing candidate counts are outside this theorem's scope.
