# 07 — Audit a feature matrix
Compare exact isotropic-Gaussian predictions with one seeded projection of a concrete matrix.

```python
import numpy as np
from rpgeom import audit
rng = np.random.default_rng(7)
X = rng.standard_normal((800, 128))
report = audit(X, m=32, q=8, n_trials=800, rng=rng)
print(report)
```

```python
payload = report.to_dict()
assert payload['r2'] > 100
assert abs(report.meas_pairwise-report.pred_pairwise) < .08
payload
```

**Interpretation.** Agreement validates the benchmark for this sample. On real embeddings, disagreement can reveal anisotropy, clustering, tails, or low intrinsic dimension and should trigger domain-specific evaluation.
