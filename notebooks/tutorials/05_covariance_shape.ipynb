# 05 — Covariance shape
Mean, overall scale, and traceless shape lose information at different rates.

```python
from rpgeom.ceilings import mean_ratio, scale_ratio, shape_ratio
d = 40
for m in (4, 8, 16, 32):
    print(m, mean_ratio(m,d), scale_ratio(m,d), shape_ratio(m,d))
```

```python
m = 8
assert shape_ratio(m,d) < mean_ratio(m,d)
print('shape / mean:', shape_ratio(m,d)/mean_ratio(m,d))
```

**Interpretation.** Diffuse scale-adjusted shape pays roughly a quadratic compression price. A finite-rank spike is a different asymptotic object and needs its own sample-size model.
