# 08 — Choose a dimension
Start from the downstream decision and solve for the smallest sketch dimension meeting its target.

```python
from rpgeom import recommend_dimension
rank_target = recommend_dimension(d=768, kendall_tau=.30)
shape_target = recommend_dimension(d=768, shape_information=.10)
print(rank_target)
print(shape_target)
```

```python
nn_target = recommend_dimension(d=768, q=10, nearest_neighbor=.35)
print(nn_target)
assert nn_target.achieved['nearest_neighbor'] >= .35
```

**Interpretation.** Different decisions imply different dimensions. Treat these Gaussian-model recommendations as transparent baselines, then audit representative data and validate the complete production pipeline.
